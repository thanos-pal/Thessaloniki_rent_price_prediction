# Creating Feature Subsets for Model Comparison

This notebook creates four different versions of the dataset, each containing a different combination of features. The purpose is to compare how model performance changes depending on the **feature set** used for training. This is a form of **ablation study**: by systematically adding or removing groups of features, we can understand which features contribute the most to predictive accuracy and whether engineered or correlated features help or hurt.

The four subsets are:

1. **Original features** — Only the columns that were present in the raw scraped dataset (26 features). This serves as the baseline, representing what we would get with no feature engineering at all.

2. **All features** — The original features plus all engineered features created in the feature engineering notebook (36 features total). This is the most feature-rich version.

3. **Uncorrelated features** — A curated subset that removes features flagged by the multicollinearity check in the train/test split notebook (26 features). This version tests whether dropping correlated predictors improves model stability.

4. **Important features** — A compact, hand-picked subset of features that were identified as the most predictive during EDA and correlation analysis (11 features). This tests whether a simpler model can achieve comparable performance with far fewer inputs.


```python
import pandas as pd
```

```python
df_train = pd.read_csv("subsets/train_predictors.csv")
df_val = pd.read_csv("subsets/val_predictors.csv")
df_test = pd.read_csv("subsets/test_predictors.csv")
```

```python
original_features = ['airConditioning', 'alarm', 'areaSqm', 'balcony', 'bathrooms',
       'closedParkingSpots', 'elevator', 'energyClass', 'fireplace', 'floor',
       'furnished', 'garden', 'hasParking', 'isNewDevelopment', 'kitchens',
       'livingRooms', 'openParkingSpots', 'petsAllowed', 'pool',
       'renovationYear', 'rooms', 'solarHeater', 'storage', 'view',
       'yearBuilt']

all_features = ['airConditioning', 'alarm', 'areaSqm', 'balcony', 'bathrooms',
       'closedParkingSpots', 'elevator', 'energyClass', 'fireplace', 'floor',
       'furnished', 'garden', 'hasParking', 'isNewDevelopment', 'kitchens',
       'livingRooms', 'openParkingSpots', 'petsAllowed', 'pool',
       'renovationYear', 'rooms', 'solarHeater', 'storage', 'view',
       'yearBuilt', 'totalParkingSpots', 'apartmentAge', 'district',
       'is_renovated', 'rooms_area', 'bath_area', 'floor_elevator',
       'floor_view']

uncorrelated_features = ['airConditioning', 'alarm', 'areaSqm', 'balcony', 'bathrooms',
       'closedParkingSpots', 'elevator', 'energyClass', 'fireplace', 'floor',
       'furnished', 'garden', 'hasParking', 'isNewDevelopment', 'kitchens',
       'livingRooms', 'petsAllowed', 'pool', 'renovationYear', 'solarHeater', 
       'storage', 'view', 'yearBuilt', 'district', 'is_renovated']

important_features = ['elevator', 'areaSqm', 'furnished', 'fireplace', 'isNewDevelopment', 'renovationYear', 'bathrooms', 
        'floor', 'district', 'storage', 'view']
```

## Saving the Subsets

For each of the four feature sets, we select the corresponding columns from the train, validation, and test DataFrames and save them to the `subsets/` directory. This produces 12 CSV files in total (4 feature sets x 3 splits).

| Feature Set | Train File | Val File | Test File | Number of Features |
|---|---|---|---|---|
| Original | `train_original.csv` | `val_original.csv` | `test_original.csv` | 26 |
| All | `train_all.csv` | `val_all.csv` | `test_all.csv` | 36 |
| Uncorrelated | `train_uncorrelated.csv` | `val_uncorrelated.csv` | `test_uncorrelated.csv` | 26 |
| Important | `train_important.csv` | `val_important.csv` | `test_important.csv` | 11 |

In the modelling notebook, we will train and evaluate the same models on each subset. This will reveal whether feature engineering, multicollinearity reduction, or aggressive feature selection leads to the best trade-off between predictive performance and model simplicity.


```python
df_train[original_features].to_csv("subsets/train_original.csv", index=False)
df_val[original_features].to_csv("subsets/val_original.csv", index=False)
df_test[original_features].to_csv("subsets/test_original.csv", index=False)

df_train[all_features].to_csv("subsets/train_all.csv", index=False)
df_val[all_features].to_csv("subsets/val_all.csv", index=False)
df_test[all_features].to_csv("subsets/test_all.csv", index=False)

df_train[uncorrelated_features].to_csv("subsets/train_uncorrelated.csv", index=False)
df_val[uncorrelated_features].to_csv("subsets/val_uncorrelated.csv", index=False)
df_test[uncorrelated_features].to_csv("subsets/test_uncorrelated.csv", index=False)

df_train[important_features].to_csv("subsets/train_important.csv", index=False)
df_val[important_features].to_csv("subsets/val_important.csv", index=False)
df_test[important_features].to_csv("subsets/test_important.csv", index=False)
```
